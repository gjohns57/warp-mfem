## Test for embedding newton viewer in notebook

```python
import sys
sys.argv = sys.argv[:1]  # strip Jupyter's --f=... kernel arg before argparse runs

from IPython.display import IFrame
from newton.examples import init, run
from mfem import MFEMExample
from concurrent.futures import ThreadPoolExecutor
executor = ThreadPoolExecutor(max_workers=1)

parser = MFEMExample.create_parser()
parser.set_defaults(viewer="viser")
viewer, args = init(parser)
example = MFEMExample(viewer, args)

print("Viewer URL:", viewer.url)
# Source - https://stackoverflow.com/q/17619964
# Posted by Julien Chastang
# Retrieved 2026-06-11, License - CC BY-SA 3.0

executor.submit(run, example, args)
IFrame("http://localhost:8081", width=800, height=600)





```
